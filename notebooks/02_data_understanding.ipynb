## Data Understanding Checklist
Before any analysis, a professional analyst checks:
1. Shape — rows and columns
2. Data types — are they correct
3. Missing values — which columns, how many
4. Duplicates — any repeated rows
5. Distributions — do values look realistic

This is called Exploratory Data Quality Assessment.
It happens before EDA. You cannot trust analysis
built on data you haven't verified.

## Data Quality Finding — TotalCharges

Found 11 rows where TotalCharges is blank.
These are NOT data entry errors.
These customers have tenure = 0, meaning they signed up
but have not yet completed their first billing cycle.

Business meaning: these are day-zero customers.
TotalCharges is blank because no bill has been issued yet.

Decision: We will handle these in the cleaning phase.
Options:
- Fill TotalCharges with 0 (they have been charged nothing)
- Remove these 11 rows (too new to analyse for churn)
We will decide which option during cleaning based on
what the analysis requires.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
print("=" * 50)
print("STEP 5A — TARGET VARIABLE: CHURN")
print("=" * 50)

# .value_counts() counts how many times each unique value appears
# This tells us how many customers churned vs stayed
churn_counts = df['Churn'].value_counts()
print(churn_counts)

# Now calculate the percentage
# This tells us what proportion of customers are churning
print("\nChurn rate as percentage:")

# len(df) gives total number of rows
# We divide each count by total rows and multiply by 100
# .round(2) rounds to 2 decimal places
print((churn_counts / len(df) * 100).round(2))
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
print("=" * 50)
print("STEP 5B — NUMERICAL COLUMN DISTRIBUTIONS")
print("=" * 50)

# .describe() gives us a statistical summary of every number column
# It shows: count, mean, min, max, and quartiles
# This tells us if values look realistic for a real business

# We select only the three numeric columns
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

# We can't run describe on TotalCharges yet because it's stored
# as text — so we convert it temporarily just for this view
# pd.to_numeric converts text to numbers
# errors='coerce' means: if a value can't be converted, make it NaN
df_temp = df.copy()
df_temp['TotalCharges'] = pd.to_numeric(
    df_temp['TotalCharges'], errors='coerce'
)

print(df_temp[numeric_cols].describe().round(2))
```

## Key Data Understanding Findings

### Churn Rate: 26.54%
Roughly 1 in 4 customers is leaving. This is the core
business problem we are solving.

### Tenure Distribution
- 25% of customers leave within 9 months
- The first 9 months = highest risk window
- Retention strategy must prioritise new customers

### Revenue Distribution
- Average customer generates $2,283 lifetime value
- Std deviation nearly equals mean — massive variation
- Budget vs premium customers behave very differently
- Customer segmentation is critical for this business

### Data Quality Issue Found
- TotalCharges stored as text, not number — must fix
- 11 rows have blank TotalCharges — tenure = 0 customers
- These will be handled in the cleaning phase

```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
print("=" * 50)
print("BUSINESS IMPACT OF 26.54% CHURN RATE")
print("=" * 50)

# Total customers in the dataset
total_customers = 7043

# Average monthly payment per customer
avg_monthly_charge = 64.76

# Total monthly revenue if every customer stays
total_monthly_revenue = total_customers * avg_monthly_charge

# Number of customers churning (26.54% of total)
# We convert percentage to decimal by dividing by 100
churned_customers = total_customers * (26.54 / 100)

# Monthly revenue lost to churn
# Each churned customer stops paying avg_monthly_charge
revenue_lost_monthly = churned_customers * avg_monthly_charge

# Annual revenue lost
# Multiply monthly loss by 12 months
revenue_lost_annually = revenue_lost_monthly * 12

print(f"Total customers          : {total_customers:,}")
print(f"Monthly revenue          : ${total_monthly_revenue:,.2f}")
print(f"Customers lost to churn  : {churned_customers:,.0f}")
print(f"Monthly revenue lost     : ${revenue_lost_monthly:,.2f}")
print(f"Annual revenue lost      : ${revenue_lost_annually:,.2f}")
```
