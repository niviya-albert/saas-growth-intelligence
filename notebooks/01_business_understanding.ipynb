# AI SaaS Growth Intelligence System
## Phase 1 — Business Understanding

**Project goal:** Identify churn patterns, predict at-risk customers,
and generate AI-powered recommendations for a SaaS business.

**Key business questions this project answers:**
1. Who is at risk of churning — and when?
2. What are the primary drivers of churn?
3. What should the business do about each at-risk customer?

## Business Understanding — Dataset Columns

### Tenure
Measures how many months a customer has stayed with the company.
High churn expected in early months. Long-tenure customers are
the loyal base and must be protected differently.

### Feature Columns (OnlineSecurity, TechSupport, etc.)
These represent product stickiness. Customers using more features
are more embedded and harder to pull away. Churned customers
likely used fewer features — suggesting a value perception problem.

### Contract Type
Month-to-month contracts have lowest switching cost = highest
churn risk. Two-year contracts force commitment. Contract type
and tenure likely correlate — two sides of the same loyalty story.
